# Baseline evaluation

## Mục tiêu
Baseline là kết quả đối chứng. mAP đo detection/keypoints trên dataset có nhãn; điểm pose trong UI đo mức giống mẫu. Hai số không thay thế nhau. Test set không dùng chọn tham số liên tục.

## Setup
Chạy từ repo đã clone ở revision bạn ghi nhận. Local dùng venv API; Colab xem docs/RESEARCH.md. Không tự cài dependency hoặc tải dữ liệu khi Run all.

```python
from pathlib import Path
import os, sys, json, tempfile
ROOT = next((p for p in [Path.cwd(), *Path.cwd().parents] if (p / "ai/research").is_dir()), None)
assert ROOT is not None, "Clone repo và chạy từ repo/notebooks"
sys.path.insert(0, str(ROOT))
os.environ["YOLO_AUTOINSTALL"] = "false"
from ai.research.cli import environment
print(json.dumps(environment(), indent=2))
```

## Các bước
Đọc cấu hình trước khi thực thi; các thao tác tốn tài nguyên mặc định tắt.

```python
from ai.research.experiments import evaluate
DATASET = None  # Path tới dataset release/dataset.yaml đã QA
CHECKPOINT = ROOT / "apps/api/models/yolov8x-pose.pt"
EXECUTE = False  # bật sau khi kiểm dataset/license/GPU
if EXECUTE:
    assert DATASET is not None
    report = evaluate(CHECKPOINT, Path(DATASET), ROOT / "ai/artifacts/baseline-r0", split="val")
    print(report)
else:
    print("NOT EXECUTED: chưa có dataset thật. Không có mAP hoặc kết luận chất lượng.")
```

Output ghi rõ thao tác thực sự chạy và thao tác bị bỏ qua. Thiếu dữ liệu không được thay bằng số giả. Khi thay dữ liệu/cấu hình, restart kernel và Run all.
